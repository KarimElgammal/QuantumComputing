```python
from braket.aws import AwsDevice
from braket.circuits import Circuit
```

```python
import os
#regions: https://docs.aws.amazon.com/braket/latest/developerguide/braket-regions.html
os.environ['AWS_PROFILE'] = 'default-us-east-1' #'default-eu-north-1' #'default-eu-north-1'#'default-us-east-1'
```

```python
device = AwsDevice("arn:aws:braket:::device/quantum-simulator/amazon/sv1")
```

```python
bell = Circuit().h(0).cnot(0, 1)
```

```python
task = device.run(bell, shots=100)
```
